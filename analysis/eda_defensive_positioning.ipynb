## Takeaways for coaches / analysts

- Organized block height sets the base shot rate: low ≫ medium > high.
- Shot phases sit in a smaller OOP bounding box — compactness correlates with danger, especially in the low block.
- On-ball pressure is the steepest contextual ladder from ~2% (no pressure) to ~21% (very high).
- Inside defensive shape and organised-defense labels separate danger; line count less so.
- Engagement type mixes volume and outcome: pressing/counter-press regain more; beaten flags predict shots.
- Team board: mean box area and width help profile which A-League sides concede more in organized defense.
